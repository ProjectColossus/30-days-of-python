## Nested Loops
- A concept that has been challenging for to understand so im practicing..
- Nested Loop is basically a Loop inside another loop(outer loop, inner loop)

    - inner Loop:

```python
#rectangle
rows = int(input())
columns = int(input())
for x in range(rows):
    for y in range(columns):
        print("x", end = " ")
    print()
```

```python
#Square
for x in range(5):
    for y in range(5):
        print("x", end =" ")
    print()
```

```python
#triangle
for x in range(1,6):
    for spaces in range(5-x):
        print(" ", end=" ")
    for stars in range(2*x -1):
        print("*", end=" ")
    print()
```

```python
#right angled
for x in range(7):
    for y in range(x):
        print("x", end=" ")
    print()
```

```python
for x in range(7):
    for y in range(7-x):
        print("x", end=" ")
    print()
```

```python
for x in range(7):
    for space in range(7-x):
        print(" ", end = " ")
    for star in range(x):
        print("x", end=" ")
    print()
```

```python
for x in range(7):
    for space in range(x):
        print(" ", end = " ")
    for star in range(7-x):
        print("x", end=" ")
    print()
```

```python
#hollow square
for x in  range(1,6):
    for y in range(1,6):
        if x==1 or x==5:
            print("x", end=" ")
        elif y ==1 or y ==5:
            print("x", end = " ")
        else:
            print(" ", end = " ")
    print()
```

```python
#floyds triangle
n = 5 #rows
num = 1 #start at 1 and keep going(increasing)
for x in range(1,n+1): 
    for y in range(x): #loops till the statisfied
        print(num, end=" ")
        num +=1 #increases with each iteration and stores them
    print() #prints in the next line
```
