```python
#7
#Declare a list variable named it_companies and assign initial values Facebook, Google, Microsoft, Apple, IBM, Oracle and Amazon.
it_companies = ['Facebook','Google','Microsoft','Apple','IBM','Oracle','Amazon']

#Print the list using print()
print(it_companies)

#Print the number of companies in the list
print(len(it_companies))

#Print the first, middle and last company
print(it_companies[0])
print(it_companies[3])
print(it_companies[6])

#Print the list after modifying one of the companies
it_companies[2] = 'Netflix'
print(it_companies)
print(len(it_companies) - 1) #use this when the list is too big

#Add an IT company to it_companies
#append() -To add item to the end of an existing list we use the method append().
it_companies.append('Tesla')
print(it_companies)

#Insert an IT company in the middle of the companies list
#insert(index, item) - u gotta tell where to to place and what item to place using the insert
it_companies.insert(4 , 'SpaceX')
print(it_companies)

#Change one of the it_companies names to uppercase (IBM excluded!)
#  variable[].upper()
it_companies[4] = it_companies[4].upper()
print(it_companies)

#Join the it_companies with a string '#;  '
it_companies = '#; '.join(it_companies)
print(it_companies)

#Check if a certain company exists in the it_companies list. 
#.index(item)
it = ['Facebook','Google','Microsoft','Apple','IBM','Oracle','Amazon']
print(it.index('Microsoft'))
print(it_companies.index('Netflix'))
```

```python
#Excercise 2 
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]

#Sort the list and find the min and max age
ages.sort()
print(ages)
min_age = ages[0] #min age
print(f"min_age = ",min_age)
max_age = ages[-1] #max age
print(f"max_age = ", max_age)
#Add the min age and the max age again to the list
ages.extend([min_age,max_age])
print(ages)

#Find the median age (one middle item or two middle items divided by two)
n = len(ages) #12
middle1 = ages[(n//2) + 1] #why we use //(floor divison) is because we want the number to be an integer to figure out the indexs if normal division causes float answer gives error
print(middle1)
middle2 = ages[(n//2)]
print(middle2)
median = (middle1 + middle2)/2
print(median)

#Find the median age (one middle item or two middle items divided by two)
Average_ages = sum(ages)/n
print("Average age:",Average_ages)

#Find the range of the ages (max minus min)
range_age = max_age - min_age
print("Range:",range_age)

#Compare the value of (min - average) and (max - average), use abs() method
#abs() - returns absolute value (basically it turns negative to positive; postive and zero remains the same)
diff_min = abs(min_age - Average_ages)
diff_max = abs(max_age - Average_ages)
print(diff_min)
print(diff_max)
```
