```python
#this is examples 
#\n - line break 
print("you are the most beautiful person i have met \nreally")
#\\ - basically helps u write the backslash 
print("this is called a backslash symbol(\\)")
#\t- adds a tab space or 4 spaces
print('days\ttopics\texercise')
print('1\t1-5\t1-10')
print('2\t5-10\t10-20')
#\- to basically show double quotes in the output
print('this is actually fun \"Hello, World\"')
```

```python
#old style string formatting 
# Strings only
first_name = 'Krishna'
last_name = 'M'
language = 'Python'
formated_string = 'I am %s %s. I teach %s' %(first_name, last_name, language)
print(formated_string)

# Strings  and numbers
radius = 10
pi = 3.14
area = pi * radius ** 2
result = 'The area of circle with a radius %d is %.2f.' %(radius, area) # 2 refers the 2 significant digits after the point
print(result)
python_libraries = ['Django', 'Flask', 'NumPy', 'Matplotlib','Pandas']
formated_string = 'The following are python libraries:%s' % (python_libraries)
print(formated_string) # "The following are python libraries:['Django', 'Flask', 'NumPy', 'Matplotlib','Pandas']"
```

```python
#new style of String formatting (.format is introduced)

a = 4
b = 3

print('{} + {} = {}'.format(a, b, a + b))
print('{} - {} = {}'.format(a, b, a - b))
print('{} * {} = {}'.format(a, b, a * b))
print('{} / {} = {:.2f}'.format(a, b, a / b)) # limits it to two digits after decimal
print('{} % {} = {}'.format(a, b, a % b))
print('{} // {} = {}'.format(a, b, a // b))
print('{} ** {} = {}'.format(a, b, a ** b))
```
